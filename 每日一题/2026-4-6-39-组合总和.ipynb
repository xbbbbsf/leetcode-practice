回溯（Backtracking）是一种通过**系统性地探索所有可能解**来解决问题的算法思想。你可以把它想象成一个**“试错”和“回退”**的过程。

它的核心逻辑是：从一个初始状态出发，沿着一条路径不断探索。如果发现当前路径走不通，或者已经找到了一个解，就**退回上一步**，尝试其他的可能性，直到所有路径都被探索完毕。

### 🧠 核心思想：决策树与深度优先搜索

解决一个回溯问题，本质上就是在遍历一棵**“决策树”**。
*   **树的根节点**：是问题的初始状态。
*   **树的分支**：代表每一步可以做的选择。
*   **树的叶子节点**：代表最终的一个结果（可能是有效解，也可能是死胡同）。

回溯算法采用**深度优先搜索（DFS）**的策略来遍历这棵树，它会一条路走到黑，直到撞了南墙再回头。

### 🛠️ 回溯三要素

一个完整的回溯算法通常包含三个核心部分，你可以把它看作一个通用模板：

1.  **路径 (Path)**：记录你已经做出的选择。
2.  **选择列表 (Choices)**：你当前可以做的所有选择。
3.  **结束条件 (Base Case)**：何时应该停止探索，收集结果。

#### 通用代码框架

```python
result = [] # 存放所有最终结果

def backtrack(路径, 选择列表):
    # 1. 结束条件：找到一个有效解
    if 满足结束条件:
        result.append(路径的副本)
        return

    # 2. 遍历所有可能的选择
    for 选择 in 选择列表:
        # 3. 剪枝：如果选择不合法，直接跳过
        if not 满足约束条件(选择):
            continue

        # --- 做出选择 ---
        将选择加入路径

        # --- 递归探索 ---
        # 基于当前选择，继续探索下一步
        backtrack(新路径, 新选择列表)

        # --- 撤销选择 (回溯的核心) ---
        # 探索完这条路后，必须撤销选择，回到之前的状态
        # 以便尝试下一个选择
        将选择从路径中移除
```

### 🧩 适用场景

回溯算法通常用于解决需要**枚举所有可能性**的组合问题。常见的题型包括：

*   **组合问题**：从一个集合中找出所有满足条件的组合。
    *   *例如*：组合总和、子集生成。
*   **排列问题**：找出一个集合的所有排列方式。
    *   *例如*：全排列。
*   **棋盘问题**：在棋盘上按规则放置棋子。
    *   *例如*：N皇后问题、解数独。
*   **路径搜索问题**：在网格或图中寻找所有可行路径。
    *   *例如*：单词搜索。

### 🌳 一个生动的比喻：走迷宫

想象你在一个迷宫里，目标是找到所有出口。

1.  **前进**：你从一个岔路口出发，选择一条路往前走（**做出选择**）。
2.  **深入**：遇到下一个岔路口，再选一条路继续深入（**递归探索**）。
3.  **碰壁**：如果走进了一条死胡同，你发现走不通了（**不满足约束条件**）。
4.  **回溯**：于是你原路返回到上一个岔路口（**撤销选择**），然后尝试另一条还没走过的路。
5.  **找到出口**：如果你走到了出口，就记录下这条路线（**满足结束条件**），然后同样返回上一个岔路口，继续寻找其他可能的出口。

这个过程不断重复，直到你尝试了所有可能的路径。

### ⚡ 性能与优化：剪枝

回溯算法本质上是暴力穷举，时间复杂度通常很高（如指数级）。因此，**剪枝（Pruning）**至关重要。

*   **什么是剪枝？** 在探索过程中，如果能提前判断某条路径不可能通向有效解，就果断放弃它，不再深入。
*   **如何剪枝？** 通过添加约束条件。例如，在“组合总和”问题中，如果当前选择的数字已经让总和超过了目标值，那么这条路径就可以直接放弃，无需再继续尝试。

总而言之，回溯是一种强大的、系统性的搜索方法，它通过“尝试-回退”的机制，能够解决许多复杂的组合问题。


````python
class Solution:
    def combinationSum(self, candidates: List[int], target: int) -> List[List[int]]:
        result = []
        # 排序是为了方便剪枝：如果当前数字已经超了，后面的肯定也超
        candidates.sort()
        
        def backtrack(start_idx, current_combination, current_sum):
            # 1. 终止条件：如果当前总和等于 target，找到一个解
            if current_sum == target:
                result.append(current_combination[:]) # 复制当前组合
                return
            
            # 2. 遍历候选数字
            for i in range(start_idx, len(candidates)):
                num = candidates[i]
                
                # 3. 剪枝：如果加上当前数字就超了，后面的也不用试了
                if current_sum + num > target:
                    break
                
                # 4. 做出选择
                current_combination.append(num)
                
                # 5. 递归探索
                # 注意：这里是 i，不是 i+1！因为同一个数字可以重复选
                backtrack(i, current_combination, current_sum + num)
                
                # 6. 撤销选择 (回溯)
                current_combination.pop()
        
        backtrack(0, [], 0)
        return result
````
