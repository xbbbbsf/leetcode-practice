```python
#去重很重要，指针的比对很重要
def threeSum(nums):
    nums.sort()  # 先排序
    result = []
    
    for i in range(len(nums)):
        # 1. 去重：如果当前数和上一个数一样，跳过，避免重复解
        if i > 0 and nums[i] == nums[i-1]:
            continue
            
        # 2. 定义双指针 (关键：在 for 循环内部定义！)
        left = i + 1
        right = len(nums) - 1
        
        # 3. 开始夹逼
        while left < right:
            current_sum = nums[i] + nums[left] + nums[right]
            
            if current_sum < 0:
                # 和太小了，把左指针右移（换一个大一点的数）
                left += 1
            elif current_sum > 0:
                # 和太大了，把右指针左移（换一个小一点的数）
                right -= 1
            else:
                # 找到了一组解！
                result.append([nums[i], nums[left], nums[right]])
                
                # 4. 找到解后，左右指针都要动
                left += 1
                right -= 1
                
                # 5. 去重：跳过所有重复的数，避免答案重复
                while left < right and nums[left] == nums[left - 1]:
                    left += 1
                while left < right and nums[right] == nums[right + 1]:
                    right -= 1
                    
    return result

# 测试一下
print(threeSum([-1, 0, 1, 2, -1, -4]))
```
