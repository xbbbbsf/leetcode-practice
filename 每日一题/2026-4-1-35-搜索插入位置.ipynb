```python
class Solution:
    def searchInsert(self, nums: List[int], target: int) -> int:
        left, right = 0, len(nums) - 1
        
        while left <= right:
            mid = (left + right) // 2
            if nums[mid] == target:
                return mid  # 找到了，直接返回
            elif nums[mid] < target:
                left = mid + 1
            else:
                right = mid - 1
        
        # 没找到，left 就是插入位置
        return left
```

如果mid大了，就得选择左边，也就是移动right，把right变为mid-1,mid小了同理
