```python
def generateParenthesis(n):
    result = []  # 用来收集所有叶子节点的结果

    # 这个函数就代表了“树上的一个节点”
    # 每次调用它，就等于走到了树的新的一层
    def backtrack(current_str, open_count, close_count):
        # 1. 终止条件 (到达叶子节点)
        if len(current_str) == 2 * n:
            result.append(current_str)
            return 

        # 2. 分支一：尝试放左括号 (向左子树走)
        # 规则：只要左括号没用完，就能放
        if open_count < n:
            # 注意：这里没有 "撤销" 操作，因为我们是传递新字符串 current_str + "("
            # 下一层递归结束后，自动回到当前层，相当于回溯了
            backtrack(current_str + "(", open_count + 1, close_count)

        # 3. 分支二：尝试放右括号 (向右子树走)
        # 规则：只有 右 < 左 时，才能放
        if close_count < open_count:
            backtrack(current_str + ")", open_count, close_count + 1)

    # 从根节点开始 (空串，0个左，0个右)
    backtrack("", 0, 0)
    
    return result
```
