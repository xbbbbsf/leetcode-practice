```python
def isValidSudoku(board):
    # 1. 创建三个“记录板”
    rows = [set() for _ in range(9)]
    cols = [set() for _ in range(9)]
    boxes = [set() for _ in range(9)]

    # 2. 遍历整个棋盘
    for i in range(9):
        for j in range(9):
            num = board[i][j]
            if num == '.':
                continue
            
            # 3. 计算九宫格索引
            box_index = (i // 3) * 3 + j // 3
            
            # 4. 检查是否重复
            if (num in rows[i] or 
                num in cols[j] or 
                num in boxes[box_index]):
                return False
            
            # 5. 将数字记录到对应的“档案”中
            rows[i].add(num)
            cols[j].add(num)
            boxes[box_index].add(num)
            
    return True
```
