```python
def romanToInt(s: str) -> int:
    # 1. 写全映射表
    map = {'I': 1, 'V': 5, 'X': 10, 'L': 50, 'C': 100, 'D': 500, 'M': 1000}
    
    result = 0
    n = len(s)
    
    # 2. 循环要包含最后一个字符
    for i in range(n): 
        # 3. 先获取当前值
        curr = map[s[i]]
        
        # 4. 关键：判断是否是最后一个字符，防止越界
        # 如果不是最后一个，且当前值 < 下一个值，则减去当前值
        if i < n - 1 and curr < map[s[i + 1]]:
            result -= curr
        else:
            # 如果是最后一个，或者当前值 >= 下一个值，则加上当前值
            result += curr
            
    return result
```

```python
#方案1
def romanToInt(s: str) -> int:
    map = {'I': 1, 'V': 5, 'X': 10, 'L': 50, 'C': 100, 'D': 500, 'M': 1000}
    result = 0
    max_val = 0  # 记录右侧出现过的最大数值
    
    # 从右往左遍历
    for char in reversed(s):
        val = map[char]
        # 如果当前值 >= 右侧最大值，说明是正常的大数，加上它
        # 如果当前值 < 右侧最大值，说明是小的减法前缀，减去它
        if val >= max_val:
            result += val
            max_val = val
        else:
            result -= val
            
    return result
```
