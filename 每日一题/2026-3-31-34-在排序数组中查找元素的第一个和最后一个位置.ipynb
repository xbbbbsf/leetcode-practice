不是双指针而是双二分查找，但是因为是排序数组所以双指针也是可以的吧？

```python
class Solution:
    def searchRange(self, nums: List[int], target: int) -> List[int]:
        def findLeftBound():
            left, right = 0, len(nums) - 1
            left_bound = -1
            while left <= right:
                mid = (left + right) // 2
                if nums[mid] == target:
                    left_bound = mid  # 记录位置
                    right = mid - 1   # 关键：继续向左找
                elif nums[mid] < target:
                    left = mid + 1
                else:
                    right = mid - 1
            return left_bound

        def findRightBound():
            left, right = 0, len(nums) - 1
            right_bound = -1
            while left <= right:
                mid = (left + right) // 2
                if nums[mid] == target:
                    right_bound = mid # 记录位置
                    left = mid + 1    # 关键：继续向右找
                elif nums[mid] < target:
                    left = mid + 1
                else:
                    right = mid - 1
            return right_bound

        # 分别查找左右边界
        left = findLeftBound()
        # 如果左边界都没找到，说明target不存在，直接返回[-1, -1]
        if left == -1:
            return [-1, -1]
        
        right = findRightBound()
        return [left, right]
```
