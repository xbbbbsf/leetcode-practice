```python
#这个是数学方法，不像栈一样占用额外空间
class Solution:
    def reverse(self, x: int) -> int:
        INT_MIN, INT_MAX = -2**31, 2**31 - 1
        result = 0
        
        # 记录符号，转成正数
        sign = -1 if x < 0 else 1
        x = abs(x)
        
        while x != 0:
            pop = x % 10
            x //= 10
            
            # 溢出检查（正数范围）
            if result > INT_MAX // 10 or (result == INT_MAX // 10 and pop > 7):
                return 0
            
            result = result * 10 + pop
        
        result *= sign
        
        # 最后检查负数溢出
        return result if INT_MIN <= result <= INT_MAX else 0
```
