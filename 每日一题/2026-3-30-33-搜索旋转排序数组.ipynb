**二分查找**（Binary Search），也叫**折半查找**，是一种非常高效的搜索算法。

简单来说，它的核心思想就是**“每次排除掉一半不可能的选项”**。

为了让你直观地理解，我们可以把它和普通的查找做个对比，并用一个生活中的例子来说明。

### 🎮 生活中的例子：猜数字游戏

假设我心里想了一个 **1 到 100** 之间的整数，你要猜出这个数字是多少。

*   **普通查找（笨办法）**：
    你从 1 开始猜：是 1 吗？不是。是 2 吗？不是……这样一个个试，运气最差的话，你要猜 100 次才能猜到。

*   **二分查找（聪明办法）**：
    你每次都用**排除法**：
    1.  你先猜中间数 **50**。
        *   我说：“**大了**”。
        *   你立刻知道：答案肯定在 1 到 49 之间，50 到 100 这一半全都不用看了。
    2.  你再猜剩下范围的中间数 **25**。
        *   我说：“**小了**”。
        *   你立刻知道：答案肯定在 26 到 49 之间。
    3.  你再猜中间数……

**结果**：每次猜测都能把范围缩小一半。对于 100 个数，最多只需要猜 **7 次**就能找到答案。如果是 100 万个数，普通查找最坏要猜 100 万次，而二分查找只需要猜 **20 次**左右。

---

### 🛠️ 算法原理（怎么操作？）

在计算机里，二分查找通常用于在一个**已经排好序的数组**（比如从小到大排列）中查找某个数。

假设数组是 `[1, 3, 5, 7, 9, 11]`，我们要找 `7`。

1.  **定边界**：
    *   左指针 `left` 指向开头（索引 0，数值 1）。
    *   右指针 `right` 指向结尾（索引 5，数值 11）。
2.  **找中间**：
    *   计算中间位置 `mid`。
    *   取出中间的数 `nums[mid]`。
3.  **比大小**：
    *   **如果相等**：恭喜你，找到了！
    *   **如果中间的数比目标大**：说明目标肯定在左边，把右指针移到中间左边（`right = mid - 1`），**右半边全部扔掉**。
    *   **如果中间的数比目标小**：说明目标肯定在右边，把左指针移到中间右边（`left = mid + 1`），**左半边全部扔掉**。
4.  **循环**：
    *   重复上面的步骤，直到找到目标，或者左右指针相遇（说明没找到）。

---

### ⚠️ 核心前提（非常重要！）

二分查找有一个**硬性条件**：
**数据必须是有序的（排好序的）。**

如果数组是乱序的（比如 `[9, 1, 5, 2...]`），你就无法通过比较中间值来判断目标在左边还是右边，这时候二分查找就失效了，只能乖乖用普通查找。

### 📊 效率对比

| 算法 | 时间复杂度 | 100万个数据要查几次？ | 适用场景 |
| :--- | :--- | :--- | :--- |
| **普通查找** | O(n) | 最多 1,000,000 次 | 数据量少，或者数据无序 |
| **二分查找** | **O(log n)** | **最多约 20 次** | **数据量大，且已排序** |

### 💡 总结

二分查找就是**“分而治之”**的策略：
不要大海捞针，而是利用**有序性**，每次砍掉一半的搜索范围，从而极速定位目标。它是计算机科学中最基础、最重要的算法之一。


```python
class Solution:
    def search(self, nums: List[int], target: int) -> int:
        left, right = 0, len(nums) - 1
        
        while left <= right:
            mid = (left + right) // 2
            
            # 1. 找到了直接返回
            if nums[mid] == target:
                return mid
            
            # 2. 判断哪一半是有序的
            # 注意：这里用 <= 是为了处理 left 和 mid 重合的情况
            if nums[left] <= nums[mid]:
                # --- 左半边是有序的 ---
                # 看看 target 是不是在左半边的范围内
                if nums[left] <= target < nums[mid]:
                    right = mid - 1  # 在左边，收缩右边界
                else:
                    left = mid + 1   # 不在左边，去右边
            else:
                # --- 右半边是有序的 ---
                # 看看 target 是不是在右半边的范围内
                if nums[mid] < target <= nums[right]:
                    left = mid + 1   # 在右边，收缩左边界
                else:
                    right = mid - 1  # 不在右边，去左边
                    
        return -1
```



答案是：**不需要遍历，我们依然可以使用二分查找！**

这正是这个算法最聪明的地方。我们并不是真的在“无序”的那一半里盲目地找，而是利用了二分查找的“递归”思想。

### 🧠 核心思想：在“混乱”中寻找新的“秩序”

你之所以会觉得“另一半是无序的，是不是只能遍历”，是因为你把“无序的那一半”当成了一个整体。

但实际上，当我们把搜索范围缩小到那一半时，**我们并没有放弃二分查找**。在下一轮循环中，我们会对这片新的、更小的范围**再次应用同样的逻辑**：

1.  找到新的中点。
2.  判断**新的左半边**和**新的右半边**谁是有序的。
3.  然后重复之前的判断过程。

**关键洞察在于：一个大的“无序”区间，当你把它一分为二时，其中至少有一半会是局部有序的。**

我们不断地“切分-判断-再切分”，就像剥洋葱一样，每一次都能利用局部有序性来排除一半的可能性，直到找到目标。

---

### 🌰 举例说明：我们是如何“二分”无序区的

让我们用一个例子来看这个过程。假设数组是 `[6, 7, 0, 1, 2, 3, 4, 5]`，我们要找 `0`。

#### 第一轮：判断全局
*   `left=0`, `right=7`, `mid=3` (值为 `1`)。
*   比较 `nums[left]=6` 和 `nums[mid]=1`。因为 `6 > 1`，所以**左半边 `[6,7,0,1]` 是无序的**。
*   那么，**右半边 `[1,2,3,4,5]` 就一定是有序的**。
*   我们的目标 `0` 在有序右半边的范围 `[1, 5]` 内吗？不在。
*   **决策**：目标不在有序的右半边，所以**去无序的左半边 `[6,7,0,1]` 找**。
    *   `right = mid - 1 = 2`。

#### 第二轮：在“无序”区里再次二分
现在我们的搜索范围是 `[6, 7, 0]` (索引0到2)。这片区域在上一轮看来是“无序”的。现在我们来看看怎么处理它。

*   `left=0`, `right=2`, `mid=1` (值为 `7`)。
*   比较 `nums[left]=6` 和 `nums[mid]=7`。因为 `6 <= 7`，所以**新的左半边 `[6, 7]` 是有序的！**
*   看，**“无序”的区域被切开后，内部也出现了“有序”的部分**。
*   我们的目标 `0` 在这个新的有序区 `[6, 7]` 内吗？不在。
*   **决策**：目标不在有序的左半边，所以**去右半边找**。
    *   `left = mid + 1 = 2`。

#### 第三轮：找到目标
现在搜索范围只剩下一个元素 `nums[2]`，值为 `0`。
*   `mid = 2`。
*   `nums[mid] == target`，找到目标！

### 📌 总结

所以，整个过程是这样的：

1.  **切一刀**：找到中点 `mid`。
2.  **找秩序**：判断 `mid` 的左边还是右边是有序的。
3.  **做判断**：看目标值是否在那个有序区间内。
    *   **在**：就收缩到这个有序区间。
    *   **不在**：就收缩到另一个（看起来无序的）区间。
4.  **重复**：回到第1步，在新的、更小的区间里重复这个过程。

我们从来没有对任何区域进行遍历。我们只是在每一步都利用“至少有一半是有序的”这个特性，不断地将搜索范围减半。这就是为什么它的时间复杂度依然是高效的 `O(log n)`。
