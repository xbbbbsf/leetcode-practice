```python
from collections import Counter

def findSubstring(s, words):
    if not s or not words: return []
    
    word_len = len(words[0])       # 单个单词长度
    num_words = len(words)         # 单词个数
    total_len = word_len * num_words # 窗口总长度 (这就是你找的那个长度！)
    
    target_count = Counter(words)  # 目标：{"foo": 1, "bar": 1}
    result = []
    
    # 【关键步骤】：我们需要尝试 word_len 种不同的起始位置
    # 比如单词长3，我们就试从 0, 1, 2 开始的情况
    for i in range(word_len):
        
        # 这里开始才是真正的“滑动窗口”
        # 我们以 word_len 为步长，在字符串上滑动
        left = i
        current_count = {} # 记录当前窗口里的单词
        
        # right 指针每次跳一个单词长度
        for right in range(i, len(s) - word_len + 1, word_len):
            word = s[right : right + word_len] # 截取单词
            
            # 如果截取的单词在目标里
            if word in target_count:
                current_count[word] = current_count.get(word, 0) + 1
                
                # 如果这个单词数量超了，说明左边要收缩
                # (这是滑动窗口的精髓：维护一个合法的窗口)
                while current_count[word] > target_count[word]:
                    left_word = s[left : left + word_len]
                    current_count[left_word] -= 1
                    left += word_len # 左边也跳一个单词长度
                
                # 如果窗口长度正好等于总长度，说明找到了！
                if right - left + word_len == total_len:
                    result.append(left)
            else:
                # 如果遇到不在 words 里的单词，整个窗口作废，重置
                current_count.clear()
                left = right + word_len
                
    return result
```
