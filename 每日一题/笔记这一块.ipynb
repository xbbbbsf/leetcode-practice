```
###  哈希表 (Hash Table) 笔记

#### 1. 什么是哈希表？
哈希表（也叫散列表）是一种非常高效的数据结构，它可以根据键（Key）直接访问内存中的数据。  
它的核心思想就像一本字典，通过拼音（键）直接找到对应的汉字（值），而不需要一页一页去翻。  
在理想情况下，它的查找、插入和删除操作的时间复杂度都是 **O(1)**。

#### 2. 核心组件
- **哈希函数 (Hash Function):**  
  它的任务是将输入的“键”（Key）转换成数组中的一个索引（下标）。  
  好的哈希函数应该计算简单且能让数据均匀分布，减少冲突。  
  常见的哈希函数有：
  - **除留余数法:** `hash(key) = key % m` (m通常取质数)。
  - **直接定址法:** `hash(key) = key` (适合范围集中的数据)。

- **哈希冲突 (Hash Collision):**  
  当两个不同的键通过哈希函数计算后，映射到了同一个索引位置。  
  比如：小李和小王都映射到了位置1。  
  冲突是不可避免的，我们需要策略来解决它。

#### 3. 解决冲突的策略
- **拉链法 (链地址法):**  
  将发生冲突的元素用链表连接起来，挂在同一个索引下。  
  就像一个桶，一个桶里可以装多个球。  
  这种方法不需要严格限制表的大小，但可能会因为链表过长而影响效率。

- **线性探测法 (开放寻址法):**  
  当发生冲突时，就往下找下一个空的位置来存放数据。  
  这要求哈希表的大小必须大于数据量，依靠空位来解决碰撞。  
  （注：这种方法在删除元素时比较麻烦，通常需要做标记而不是直接移除）。

#### 4. 什么时候用哈希表？
当你需要**快速判断一个元素是否出现在集合中**时，就要想到哈希表。  
典型题目：两数之和、有效的字母异位词、快乐数。

---

###  排序算法 (Sorting) 笔记

#### 1. 为什么要排序？
排序是很多高效算法的前提。  
比如：只有在有序数组中，我们才能使用**二分查找**（时间复杂度 O(log n)）；  
或者像“三数之和”这类问题，排序后利用双指针可以大大降低时间复杂度。

#### 2. Python 中的排序方法
Python 提供了非常便捷的排序工具，我们需要熟练掌握：

- **`list.sort()`**  
  直接修改原列表，不返回新列表（就地排序）。  
  适合不需要保留原数组的场景。  
  语法：`nums.sort()`

- **`sorted(list)`**  
  返回一个新的排序后的列表，原列表保持不变。  
  适合需要保留原数据的场景。  
  语法：`new_nums = sorted(nums)`

#### 3. 常用参数
- **`reverse=True`:** 用于降序排列。  
  例如：`nums.sort(reverse=True)` 会从大到小排序。
- **`key`:** 用于指定排序的规则（比如按字符串长度排，按字典的某个键排）。

#### 4. 排序在算法题中的作用
- **去重:** 相同的元素会排在一起，方便用双指针跳过重复值。
- **剪枝:** 排序后，如果当前数已经大于目标值，后面的数就不用看了，直接跳出循环。
- **双指针:** 排序后，利用数组有序的特性，可以用左右指针从两端向中间逼近，解决很多问题（如两数之和、三数之和）。
```

审题与分析

在动手写代码之前，必须花时间彻底理解题目要求。

明确输入输出：
    输入是什么类型的数据？（整数、数组、字符串、链表、树等）
    输出需要返回什么？（布尔值、数值、数组、节点等）
理解约束条件：
    数据的规模有多大？（例如：1 <= n <= 10^5）
    时间和空间复杂度的限制是什么？
    边界条件是什么？（空数组、单个元素、负数、大数等）
识别关键信息：
    题目中是否有提示性的关键词？（例如：“排序”、“唯一”、“子序列”、“连续”等）

寻找思路与策略

这是解题的核心环节，可以尝试以下几种常用策略：

暴力法：
    最直接、最简单的思路，通常是遍历所有可能的情况。
    优点：容易想到，能保证正确性。
    缺点：时间复杂度通常较高，可能无法通过所有测试用例。
    作用：作为起点，帮助你理解问题，有时可以从中发现优化的线索。
双指针：
    使用两个指针（变量）来遍历数据结构。
    常见类型：
        快慢指针： 一个快一个慢，用于检测环、删除重复元素等。
        左右指针： 从两端向中间移动，常用于有序数组的查找（如两数之和、三数之和）。
滑动窗口：
    维护一个窗口（子数组），根据条件动态调整窗口的大小和位置。
    适用于寻找满足特定条件的最小子数组、最长子串等问题。
二分查找：
    在有序数据中，每次将搜索范围减半。
    核心思想：left, right, mid。
    适用于在有序数组中查找目标值、寻找边界等问题。
哈希表：
    利用哈希表的快速查找特性（O(1)）。
    适用于需要快速判断元素是否存在、统计频率、寻找配对等问题。
递归与分治：
    将大问题分解为小问题，递归解决小问题，再合并结果。
    适用于树、图、排序（归并排序、快速排序）等问题。
动态规划：
    适用于具有重叠子问题和最优子结构性质的问题。
    核心步骤：定义状态、确定状态转移方程、初始化、返回结果。

代码实现

确定思路后，开始编写代码。

先写伪代码： 用自然语言描述算法步骤，理清逻辑。
模块化编程： 将复杂逻辑拆分为多个小函数，提高代码可读性和可维护性。
变量命名清晰： 使用有意义的变量名（如 left, right, sum, result），避免使用单个字母（如 i, j, k 除外）。
处理边界条件： 在代码开头或关键位置，先处理特殊情况（如空输入、单个元素等）。

调试与优化

代码写完后，需要验证其正确性和效率。

手动测试： 使用简单的测试用例（包括边界条件）手动执行代码，检查输出是否符合预期。
调试工具： 利用 IDE 的调试功能（如断点、单步执行）查看变量值和程序流程。
时间复杂度分析： 分析算法的时间复杂度，看是否有优化空间。
空间复杂度分析： 检查是否使用了过多的额外空间，能否优化。

总结与复盘

解完题后，花几分钟回顾整个过程。

总结经验： 这道题用到了哪些算法思想？有哪些技巧可以复用？
记录难点： 哪里容易出错？如何避免？
寻找更优解： 是否有更简洁、更高效的解法？可以参考题解或他人的代码。
