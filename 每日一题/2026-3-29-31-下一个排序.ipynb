### 🧠 核心思想：字典序（就像查字典一样）

想象一下，你把这些数字组成的所有排列，按照**从小到大**的顺序写在纸上。
比如数字 `1, 2, 3` 的所有排列顺序是：
1.  `1, 2, 3`
2.  `1, 3, 2`  <-- 下一个排列
3.  `2, 1, 3`
4.  `2, 3, 1`
5.  `3, 1, 2`
6.  `3, 2, 1`  <-- 最大了，下一个回到 `1, 2, 3`

**数学思想就是：如何不列出所有数字，直接算出当前数字的“下一行”是什么？**

---

### 🌰 例子一：普通情况 `[1, 2, 4, 3]`

我们要找比 `1243` 大的下一个数。

#### 第一步：找“末尾”（从右往左找“下坡”）
我们从最右边往左看，看看哪里的数字开始**变小**了（也就是打破了从右往左递增的趋势）。
- 看 `3`：它是最后一个，不用看。
- 看 `4`：`4` 比 `3` 大。说明 `4, 3` 这一段已经是**倒序**（最大组合）了，没法通过交换它们变大。
- 看 `2`：**停！** `2` 比 `4` 小。
- **发现目标**：`2` 就是我们要找的位置（索引 `i`）。
    - *数学含义*：`2` 是“高位”，后面的 `4, 3` 是“低位”。要让数字变大，必须动这个 `2`。

#### 第二步：找“替身”（从右往左找“刚好比它大”的数）
我们要把 `2` 换掉。为了保持“大得最少”，我们要从后面的 `4, 3` 里，找一个**比 2 大，但又是里面最小的**数。
- 从最右边找：`3` 比 `2` 大。
- 再往左看：`4` 比 `2` 大，但 `4` 比 `3` 大，所以 `3` 才是“刚好比 2 大”的那个数。
- **交换**：把 `2` 和 `3` 交换。
- 数组变成：`[1, 3, 4, 2]`

#### 第三步：反转“后半段”（让后面变成最小）
现在数组是 `1, 3, 4, 2`。
注意，`3` 后面的 `4, 2` 依然是**倒序**的（这是算法的特性）。
为了让整体数字最小，我们需要把 `4, 2` **反转**成 `2, 4`。
- **最终结果**：`[1, 3, 2, 4]`

**验证**：`1243` 的下一个数确实是 `1324`。

### 🌰 例子二：特殊情况 `[3, 2, 1]`

我们要找比 `321` 大的下一个数。

#### 第一步：找“末尾”
从右往左看：
- `1` -> `2`：2 比 1 大。
- `2` -> `3`：3 比 2 大。
- 走到头了，也没找到“左边比右边小”的情况。
- **发现目标**：没找到（`i = -1`）。

#### 数学含义
这意味着整个数组已经是**完全倒序**了，也就是**最大**的排列。

#### 怎么办？
既然已经是最大了，下一个排列就是**最小**的排列。
- **操作**：直接反转整个数组。
- **最终结果**：`[1, 2, 3]`

### 📌 总结这个“数学思想”

这个算法其实就是在模拟我们**进位**的过程：

- **从右往左找**：就像我们做加法 `199 + 1`，我们要从个位开始看，直到找到那个可以“进位”的地方。在这里，我们找的是那个“可以被替换变大”的高位数字。
- **交换**：把那个高位数字换成一个“稍微大一点点”的数字。
- **反转**：把后面所有的数字变成“最小状态”（从小到大），这样整体增加得最少。

是不是清晰多了？你可以拿笔在纸上画一下 `[1, 3, 2]` 的下一个排列，试试看能不能推出来！

```python
class Solution:
    def nextPermutation(self, nums: List[int]) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """
        # 1. 从右往左找第一个 nums[i] < nums[i+1] 的位置
        i = len(nums) - 2
        while i >= 0 and nums[i] >= nums[i + 1]:
            i -= 1
        
        # 2. 如果找到了这样的位置（说明不是完全倒序，比如 321）
        if i >= 0:
            # 从右往左找第一个比 nums[i] 大的数
            j = len(nums) - 1
            while j >= 0 and nums[j] <= nums[i]:
                j -= 1
            # 交换这两个数
            nums[i], nums[j] = nums[j], nums[i]
        
        # 3. 反转 i+1 到末尾的部分
        # 如果 i = -1（说明原数组是完全倒序），则反转整个数组
        self.reverse(nums, i + 1, len(nums) - 1)

    def reverse(self, nums, start, end):
        while start < end:
            nums[start], nums[end] = nums[end], nums[start]
            start += 1
            end -= 1
```
