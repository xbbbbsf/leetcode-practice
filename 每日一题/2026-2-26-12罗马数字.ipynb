两种思路  
按位查表法，拆分出千位百位个位的数字再用已经准备好的表格去表示，用%与//来拆  
贪心算法，减去能减的最大数字，有点像二进制的算法？  

```python
def intToRoman(num: int) -> str:
    # 预定义每一位的映射表
    # 索引 0 代表该位为 0 (空字符串)，索引 1-9 代表对应的罗马数字
    thousands = ["", "M", "MM", "MMM"]
    hundreds  = ["", "C", "CC", "CCC", "CD", "D", "DC", "DCC", "DCCC", "CM"]
    tens      = ["", "X", "XX", "XXX", "XL", "L", "LX", "LXX", "LXXX", "XC"]
    ones      = ["", "I", "II", "III", "IV", "V", "VI", "VII", "VIII", "IX"]
    
    # 分解数字并拼接
    # num // 1000 取千位
    # (num % 1000) // 100 取百位
    # (num % 100) // 10 取十位
    # num % 10 取个位
    
    return thousands[num // 1000] + \
           hundreds[(num % 1000) // 100] + \
           tens[(num % 100) // 10] + \
           ones[num % 10]

# 测试
print(intToRoman(1994)) # 输出: MCMXCIV
print(intToRoman(58))   # 输出: LVIII
print(intToRoman(3))    # 输出: III
```

```python
def intToRoman_greedy(num: int) -> str:
    # 必须按从大到小排序，包含减法特例 (如 900, 400, 90, 40, 9, 4)
    val_map = [
        (1000, "M"), (900, "CM"), (500, "D"), (400, "CD"),
        (100, "C"),  (90, "XC"),  (50, "L"),  (40, "XL"),
        (10, "X"),   (9, "IX"),   (5, "V"),   (4, "IV"),
        (1, "I")
    ]
    
    result = []
    
    for value, symbol in val_map:
        # 如果当前数字大于等于某个值，就一直减去它，并添加对应符号
        while num >= value:
            num -= value
            result.append(symbol)
        # 如果 num 变为 0，提前结束
        if num == 0:
            break
            
    return "".join(result)

# 测试
print(intToRoman_greedy(1994)) # 输出: MCMXCIV
```
