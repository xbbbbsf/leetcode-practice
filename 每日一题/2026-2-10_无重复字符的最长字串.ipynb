```python
class Solution:
    def lengthOfLongestSubstring(self, s: str) -> int:
        left = 0
        max_len = 0
        char_set = set()  
    
        for right in range(len(s)): 
            
            
            while s[right] in char_set:  # 如果新字符已在窗口中
                char_set.remove(s[left])  # 左指针字符移出窗口
                left += 1                  # 左指针右移（收缩）
            
            # 现在窗口没有重复了，加入新字符
            char_set.add(s[right])
            max_len = max(max_len, right - left + 1)
        
        return max_len
```
