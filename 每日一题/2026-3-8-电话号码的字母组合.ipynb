```python
def letterCombinations(digits):
    if not digits:
        return []
        
    # 1. 建立哈希表
    phone_map = {
        "2": "abc", "3": "def", "4": "ghi", "5": "jkl",
        "6": "mno", "7": "pqrs", "8": "tuv", "9": "wxyz"
    }
    
    result = []
    
    # 定义递归函数
    def backtrack(index, path):
        # 2. 递归终止：如果处理完了所有数字
        if index == len(digits):
            result.append(path)
            return
            
        # 3. 单层逻辑：获取当前数字对应的字母
        current_digit = digits[index]
        letters = phone_map[current_digit]
        
        # 遍历当前数字的每一个字母
        for letter in letters:
            # 将字母加入路径
            backtrack(index + 1, path + letter)
            # 注意：这里没有显式的 pop，因为我们用的是字符串拼接，
            # path + letter 生成了新对象，原 path 没变，所以不需要回溯删除。
            # 如果用列表存储 path，这里需要 path.pop()。
    
    # 从第0个数字开始递归
    backtrack(0, "")
    return result
```
