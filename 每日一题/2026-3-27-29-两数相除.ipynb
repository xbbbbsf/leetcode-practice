```python
class Solution:
    def divide(self, dividend: int, divisor: int) -> int:
        # 1. 处理溢出边界：-2^31 / -1 = 2^31 (溢出)
        INT_MAX = 2**31 - 1
        INT_MIN = -2**31
        if dividend == INT_MIN and divisor == -1:
            return INT_MAX

        # 2. 确定符号
        # 异号为负，同号为正
        negative = (dividend < 0) != (divisor < 0)
        
        # 3. 全部转为正数处理（避免负数位移的复杂性）
        # 注意：Python 的 abs(-2**31) 还是 -2**31，所以要先转 long 或者处理一下
        # 但 Python 自带大整数，直接 abs 也可以，只要最后转回 int
        a = abs(dividend)
        b = abs(divisor)
        
        result = 0
        
        # 4. 倍增法核心逻辑
        while a >= b:
            temp = b      # 当前除数的倍数
            multiple = 1  # 倍数值（1, 2, 4...）
            
            # 尝试找到最大的 temp，使得 temp <= a
            # 这里用 temp << 1 相当于 temp * 2
            while a >= (temp << 1):
                temp <<= 1
                multiple <<= 1
            
            # 减去最大的倍数
            a -= temp
            result += multiple
            
        # 5. 加上符号
        if negative:
            result = -result
            
        # 6. 处理结果范围（虽然 Python 不会溢出，但题目要求）
        return max(INT_MIN, min(INT_MAX, result))
```
