移动高的 = 必死（宽度减小，高度上限被锁死）  
移动矮的 = 唯一希望（虽然宽度减小，但高度上限可能打开）  
核心思路：双指针，因为矮的指针决定水的高度，所以最高点是多高没有影响

```python
class Solution:
    def maxArea(self, height: List[int]) -> int:
        # 1. 初始化双指针
        left = 0                  # 最左边
        right = len(height) - 1   # 最右边
        max_area = 0              # 记录最大面积
        
        # 2. 当左指针在右指针左边时，循环继续
        while left < right:
            # A. 计算当前宽度
            width = right - left
            
            # B. 计算当前高度 (取两根柱子的较短者)
            h = min(height[left], height[right])
            
            # C. 计算当前面积
            current_area = width * h
            
            # D. 更新最大面积
            if current_area > max_area:
                max_area = current_area
            
            # E. 【核心策略】移动指针
            # 如果左边矮，就移动左边（因为留着矮的也没用，宽度还变小，必须换高的试试）
            if height[left] < height[right]:
                left += 1
            # 如果右边矮，就移动右边
            elif height[right] < height[left]:
                right -= 1
            # 如果一样高，移动哪边都行（或者两边都动）
            else:
                left += 1
                right -= 1
                
        return max_area
```
