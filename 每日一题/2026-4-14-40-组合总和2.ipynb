```python
class Solution:
    def combinationSum2(self, candidates: List[int], target: int) -> List[List[int]]:
        result = []
        # 1. 必须先排序！这是去重的前提
        candidates.sort()

        def backtrack(path, current_sum, start_index):
            if current_sum == target:
                result.append(path[:])
                return
            
            if current_sum > target:
                return

            for i in range(start_index, len(candidates)):
                # --- 核心改动：树层去重 ---
                # 如果当前数字和前一个数字相同，且前一个数字不是在这一层递归的“长辈”（即 i > start_index）
                # 说明前一个相同的数字已经被试过了（或者被回溯了），为了避免重复结果，直接跳过
                if i > start_index and candidates[i] == candidates[i-1]:
                    continue

                num = candidates[i]
                
                # 剪枝：如果加上去超了，后面的肯定也超（因为排过序）
                if current_sum + num > target:
                    break

                # 做出选择
                path.append(num)
                
                # --- 另一个改动：递归传 i + 1 ---
                # 因为题目说“每个数字只能用一次”，所以下一层必须从 i+1 开始，不能回头
                backtrack(path, current_sum + num, i + 1)
                
                #
```
