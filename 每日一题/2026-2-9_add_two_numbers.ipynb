```python
class Solution:
    def addTwoNumbers(self, l1: ListNode, l2: ListNode) -> ListNode:
        dummy = ListNode(0)     # 虚拟头节点（简化代码）
        current = dummy         # 当前正在处理的节点
        carry = 0               # 进位
        
        while l1 or l2 or carry:    # 只要还有节点或进位，就继续
            # 取值（如果节点为空，用0）
            x = l1.val if l1 else 0
            y = l2.val if l2 else 0
            
            # 计算总和
            sum = x + y + carry
            
            # 新的进位？（总和超过9就进位）
            carry = sum // 10       # 整数除法，比如14//10=1
            
            # 当前位写几？（只保留个位）
            digit = sum % 10        # 取余，比如14%10=4
            
            # 创建新节点，接在后面
            current.next = ListNode(digit)
            
            current = current.next      # current 移动到新创建的节点
            if l1: l1 = l1.next         # l1 移动到下一个（如果还有）
            if l2: l2 = l2.next         # l2 移动到下一个（如果还有）
        
        return dummy.next   # 返回结果（跳过虚拟头）
```
