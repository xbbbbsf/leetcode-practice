双指针居然不止有相对，还有快慢的，可以表示出一个“窗口”，  
而且在链表里面，不是用索引去走，而是用.next去表示

```python
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next

class Solution:
    def removeNthFromEnd(self, head: Optional[ListNode], n: int) -> Optional[ListNode]:
        # 创建虚拟头节点，简化对头节点被删除的情况的处理
        dummy_head = ListNode(next=head)
        
        # 初始化快慢指针
        fast = slow = dummy_head
        
        # 快指针先走 n 步
        for _ in range(n):
            fast = fast.next
            
        # 快慢指针同时移动，直到快指针到达链表末尾
        while fast.next:
            fast = fast.next
            slow = slow.next
        
        # 此时 slow 指针指向倒数第 n 个节点的前一个节点
        # 跳过倒数第 n 个节点
        slow.next = slow.next.next
        
        # 返回新的头节点
        return dummy_head.next
```
