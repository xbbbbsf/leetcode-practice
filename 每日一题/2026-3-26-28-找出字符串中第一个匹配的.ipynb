暴力解法：

```python
class Solution:
    def strStr(self, haystack: str, needle: str) -> int:
        len_h = len(haystack)
        len_n = len(needle)
        
        # 边界情况：如果 needle 是空字符串，返回 0 (题目通常这样定义)
        if len_n == 0:
            return 0
            
        # 外层循环：i 是 haystack 的指针 (你的 slow)
        # 只需要遍历到 len_h - len_n 即可，后面不够长了
        for i in range(len_h - len_n + 1):
            
            # 内层循环：j 是 needle 的指针 (你的 fast)
            # 我们假设从 i 开始匹配，看看能不能匹配完整个 needle
            match = True
            for j in range(len_n):
                if haystack[i + j] != needle[j]:
                    match = False
                    break # 只要有一个不对，就不用比了，跳出内层循环
            
            # 如果内层循环顺利跑完，说明全部匹配
            if match:
                return i
                
        # 如果都没找到
        return -1
```
