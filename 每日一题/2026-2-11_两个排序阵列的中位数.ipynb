第一次思考，排序以后再取中位数：用排序生成一个有序数列，然后在用len来取中位数，再来个if循环使其中位数不是整数的的情况下计算中位数

```python
class Solution:
    def findMedianSortedArrays(self, nums1, nums2):
        merged = sorted(nums1 + nums2)#此处的排序是sorted，Python 的 sort() 和 sorted() 使用的是 Timsort，这是一种混合排序算法
        n = len(merged)
        if n % 2 == 1:
            return merged[n // 2]
        else:
            return (merged[n // 2 - 1] + merged[n // 2]) / 2
```

注：sort与sorted的区别

```python
nums = [3, 1, 4, 1, 5]

# sorted() - 返回新列表，原列表不变
new_nums = sorted(nums)
print(nums)      # [3, 1, 4, 1, 5]  ← 原列表不变
print(new_nums)  # [1, 1, 3, 4, 5]  ← 新列表有序

# .sort() - 原地排序，返回 None
nums.sort()
print(nums)      # [1, 1, 3, 4, 5]  ← 原列表被修改
```

法2：双指针排序法：利用了数据本身就排序好了的情况下

```python
def findMedianSortedArrays(nums1, nums2):
    # 双指针合并两个有序数组
    merged = []
    i = 0  # nums1 的指针
    j = 0  # nums2 的指针
    
    # 当两个数组都还有元素时，逐个比较
    while i < len(nums1) and j < len(nums2):
        if nums1[i] < nums2[j]:
            # nums1 的当前元素更小，先放入结果
            merged.append(nums1[i])
            i += 1  # nums1 指针右移
        else:
            # nums2 的当前元素更小（或相等），先放入结果
            merged.append(nums2[j])
            j += 1  # nums2 指针右移
    
    # 其中一个数组放完了，另一个可能还有剩余
    # 直接追加剩余部分（因为已经有序）
    if i < len(nums1):
        merged.extend(nums1[i:])  # nums1 剩余部分
    
    if j < len(nums2):
        merged.extend(nums2[j:])  # nums2 剩余部分
    
    # 现在 merged 是有序的，找中位数
    n = len(merged)
    if n % 2 == 1:
        return merged[n // 2]
    else:
        return (merged[n // 2 - 1] + merged[n // 2]) / 2
```
