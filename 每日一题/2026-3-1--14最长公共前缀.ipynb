14题，我感觉比较简单，用指针比对就行了，但是问题是可能会出现多个比对目标，所以可能需要变动数量的指针，所以不能用指针？而是直接索引比对？先get strs这个列表的len，然后把这些字符串依索引比对，将=的情况放入result这样可以吗？

```python
def longestCommonPrefix(strs):
    if not strs:
        return ""
    
    result = ""
    # 获取列表长度
    num_strings = len(strs)
    
    # 找一个基准长度，通常取第一个字符串的长度，防止后面索引越界
    # (其实也可以用 min(len(s) for s in strs)，但那样要遍历一遍，不如直接在循环里判断)
    
    # 外层循环：遍历索引 i (纵向扫描)
    # 这里我们假设第一个字符串是最短的，如果其他字符串更短，会在内层循环里被检测到
    for i in range(len(strs[0])): 
        # 取出当前要比较的字符（以第一个字符串为基准）
        current_char = strs[0][i]
        
        # 内层循环：遍历每一个字符串 (除了第一个，因为它已经被当作基准了)
        for j in range(1, num_strings):
            # 关键判断：
            # 1. 如果当前字符串 j 的长度不够了 (i 超出了它的范围)【我没想到这种情况】
            # 2. 或者当前字符不匹配
            if i >= len(strs[j]) or strs[j][i] != current_char:
                return result # 直接返回当前的结果，不再继续
        
        # 如果内层循环顺利走完，说明所有字符串的第 i 个字符都匹配
        result += current_char
    
    return result
```
