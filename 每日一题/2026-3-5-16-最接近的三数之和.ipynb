#我的思路  
16题，最接近的三数之解，我感觉很像15题，我觉得也可以双指针来做，先排序，再双指针，小了或者大了就移，不过问题是移哪两个，我认为可以先遍历一遍，最接近的先固定下来，剩下两个用双指针确定，还得记得查重

ai生成的代码和我的区别：我没想到可以在加上closest_sum = current_sum这一步来更新最小的值，整体上还是跟15题很像

```python
def threeSumClosest(nums, target):
    nums.sort()  # 1. 先排序
    n = len(nums)
    closest_sum = float('inf')  # 初始化为无穷大

    for i in range(n - 2):  # 2. 遍历固定第一个数
        # 3. 查重：如果当前数和前一个数一样，跳过
        if i > 0 and nums[i] == nums[i - 1]:
            continue

        left = i + 1
        right = n - 1

        while left < right:
            current_sum = nums[i] + nums[left] + nums[right]

            # 4. 更新最接近的值
            if abs(current_sum - target) < abs(closest_sum - target):
                closest_sum = current_sum

            # 5. 核心移动逻辑
            if current_sum < target:
                left += 1  # 和太小了，左指针右移（增大数值）
            elif current_sum > target:
                right -= 1  # 和太大了，右指针左移（减小数值）
            else:
                return current_sum  # 完全相等，直接返回

    return closest_sum
```
