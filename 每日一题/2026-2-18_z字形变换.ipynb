```python
class Solution:
    def convert(self, s: str, numRows: int) -> str:
    
        if numRows == 1 or numRows >= len(s):
            return s
        
        # 创建numRows个空字符串
        rows = [''] * numRows
        row, direction = 0, 1
        
        for char in s:
            rows[row] += char
            # 到达边界时改变方向
            if row == 0:
                direction = 1
            elif row == numRows - 1:
                direction = -1
            row += direction
        
        return ''.join(rows)
```

```python
rows = ['', '', '']  # 三个空字符串
row = 0
char = 'P'

rows[row] += char   # rows[0] 变成 'P'

char = 'A'
rows[row] += char   # rows[0] 变成 'PA'
#效果：把字符一个个接到字符串后面。
```
