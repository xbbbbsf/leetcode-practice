链表的互相交换有点恶心

```python
class Solution:
    def swapPairs(self, head: Optional[ListNode]) -> Optional[ListNode]:
        # 1. 创建哨兵节点，指向头节点 (解决头节点变换的麻烦)
        dummy = ListNode(0)
        dummy.next = head
        
        # 2. 初始化 prev 指针，它永远指向“待交换对”的前一个节点
        prev = dummy
        
        # 3. 循环条件：只要后面还有至少两个节点，就继续交换
        while prev.next and prev.next.next:
            # 标记出要交换的两个节点
            first = prev.next       # 第一个节点 (比如 1)
            second = prev.next.next # 第二个节点 (比如 2)
            
            # --- 开始交换三部曲 ---
            
            # 第一步：prev 指向 second (前驱连老二)
            # 结果：dummy -> 2
            prev.next = second
            
            # 第二步：first 指向 second 的下一个 (老大连老三)
            # 结果：1 -> 3 (此时 1 已经脱离组织了)
            first.next = second.next
            
            # 第三步：second 指向 first (老二连老大)
            # 结果：2 -> 1
            second.next = first
            
            # --- 交换结束，现在是：prev -> 2 -> 1 -> 3 ---
            
            # 4. 移动 prev 指针
            # 现在的 first (也就是 1) 变成了这一组的尾巴，下一轮它就是前驱了
            prev = first 
            
        return dummy.next
```

递归  
head.next = self.swapPairs(newHead.next)  
这行特别好，相当于信任后面的递归，把递归得出的结果接在了他后面

```python

class Solution:
    def swapPairs(self, head: ListNode) -> ListNode:
        if not head or not head.next:
            return head
        newHead = head.next
        head.next = self.swapPairs(newHead.next)
        newHead.next = head
        return newHead

# 作者：力扣官方题解
# 链接：https://leetcode.cn/problems/swap-nodes-in-pairs/solutions/444474/liang-liang-jiao-huan-lian-biao-zhong-de-jie-di-91/
# 来源：力扣（LeetCode）
# 著作权归作者所有。商业转载请联系作者获得授权，非商业转载请注明出处。
```
