输入：数组 nums，目标值 target
输出：两个数的下标

步骤：
1. 遍历所有结果，找到能实现target的那一个结果
2. 将找到的结果为ture，if...ture使其输出
3. 把那个结果转化为需要的索引的结果
疑问：第二点我不知道怎么实现，第三点我不知道怎么转化为索引

```python
#全遍历一遍
class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        seen = {}  # 值 -> 索引
        
        for i, num in enumerate(nums):
            need = target - num      # 需要找的另一个数
            if need in seen:         # 之前见过吗？
                return [seen[need], i]  # 返回 [之前的索引, 现在的索引]
            seen[num] = i            # 存入当前数
        
        return []
```

```python
#哈希
class SolutionOptimized:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        seen = {} # 创建一个空字典（哈希表）
        for i, num in enumerate(nums):# enumerate(nums)	同时拿到索引和值 i是索引，num是值
            need = target - num  # 需要找的另一个数
            if need in seen: # 在哈希表里查 need（O(1)时间！）
                return [seen[need], i] # 找到了！返回 [之前的索引, 现在的索引]
            seen[num] = i  # 把当前数存入哈希表 用来给上面的循环提供“素材”

```

复盘总结
 
    关键点
    1. j 从 i+1 开始: 避免重复和自加
    2. 哈希表核心: 用空间换时间，O(1) 查找
    3. Python dict: 就是哈希表，插入/查找都是 O(1)
   
   易错点
    ❌ `range(i)` 也可以，但 `range(i+1, n)` 更直观（往后找）
    ❌ 不要写成 `c -= 1`，要写成 `count[c] -= 1`（改的是值，不是 key）（这个是哈希表的错的点）
