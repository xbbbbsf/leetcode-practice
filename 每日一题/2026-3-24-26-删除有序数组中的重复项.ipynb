数组是怎么删除的？  
题目里好像不是直接删除  
感觉和列表很像

```python
# 处理边界情况（如果数组为空，直接返回 0）
if not nums:
    return 0

slow = 0
# 快指针从 1 开始遍历直到末尾
for fast in range(1, len(nums)):
    
    # 核心判断
    if nums[fast] != nums[slow]:
        # 1. 慢指针先挪窝
        slow += 1
        # 2. 把快指针的值搬过来（覆盖掉旧的/重复的值）
        nums[slow] = nums[fast]
    
    # 如果相等（else），什么都不做，fast 会在 for 循环里自动 +1
    # 这就是“跳过/删除”重复项的操作

# 最后返回长度（下标 + 1）
return slow + 1
```
