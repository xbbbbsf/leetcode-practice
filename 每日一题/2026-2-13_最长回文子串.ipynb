第5题：最长回文子串 - 学习总结

题目

找到字符串中最长的回文子串。

我的思路演变

初版：遍历所有子串，判断回文
问题：O(n³) 太慢
改进：第一遍找最大长度，第二遍找位置
问题：没必要两次遍历
最终：中心扩展，边找边记录
优点：O(n²)，一次搞定

关键突破：中心扩展法
回文 = 中心对称
从每个字符（或字符间）向两边扩展：

奇数长度：中心是一个字符 expand(i, i)
偶数长度：中心在两个字符间 expand(i, i+1)

```python
class Solution:
    def longestPalindrome(self, s: str) -> str:
        # 边界情况
        if not s or len(s) < 1:
            return ""
        
        start = 0      # 最长回文的起始位置
        max_len = 0    # 最长回文的长度
        
        # 遍历每个可能的中心
        for i in range(len(s)):
            # 奇数长度，中心是一个字符
            len1, start1 = self.expand(s, i, i)
            
            # 偶数长度，中心在两个字符中间
            len2, start2 = self.expand(s, i, i + 1)
            
            # 取较大的更新记录
            if len1 > max_len:
                max_len = len1
                start = start1
            
            if len2 > max_len:
                max_len = len2
                start = start2
        
        # 返回子串
        return s[start : start + max_len]
    
    def expand(self, s: str, left: int, right: int) -> (int, int):
        while left >= 0 and right < len(s) and s[left] == s[right]:
            left -= 1
            right += 1
        
        length = right - left - 1
        start = left + 1
        
        return length, start
```
