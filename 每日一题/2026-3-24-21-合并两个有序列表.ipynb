# 📝 链表核心知识点笔记

### 1. ❌ 链表没有索引 (No Index)
*   **数组 (Array/List)**：像**电影院座位**，有编号（0, 1, 2...）。可以直接跳过去 `list[5]`，速度极快 ($O(1)$)。
*   **链表 (Linked List)**：像**寻宝游戏**，**没有编号**。
    *   想找第 5 个节点？必须从第 1 个开始，顺着 `next` 指针一个个走过去 ($O(N)$)。
    *   **结论**：链表**不能**用 `list[i]` 访问，只能靠指针移动。

### 2. 🔍 `.val` vs 节点对象
*   **`node.val`**：节点里存的**数据**（如数字 `5`）。仅用于**比较大小**或计算。
*   **`node`**：**节点本身**（包含数据和 `next` 指针）。用于**连接**和**移动**。
*   **铁律**：
    *   比大小：用 `.val` (`if l1.val < l2.val`)
    *   连链表：用节点 (`tail.next = l1`)
    *   移指针：用 `.next` (`l1 = l1.next`)

### 3. 🛡️ 虚拟头节点 (Dummy Head)
*   **作用**：消除“第一个节点”的特殊处理逻辑。
*   **用法**：
    1.  创建 `dummy = ListNode(0)`。
    2.  让 `tail` 从 `dummy` 开始。
    3.  最后返回 `dummy.next`（真正的头）。
*   **好处**：代码逻辑统一，无需判断 `if head is None`。

### 4. 🏃 双指针移动模板
```python
dummy = ListNode(0)
tail = dummy

while l1 and l2:
    if l1.val < l2.val:
        tail.next = l1      # 1. 连节点
        tail = tail.next    # 2. 移 tail (关键！)
        l1 = l1.next        # 3. 移 l1
    else:
        tail.next = l2
        tail = tail.next
        l2 = l2.next

# 接上剩余部分
tail.next = l1 if l1 else l2

return dummy.next
```

### 5. ⚡ 复杂度
*   **时间**：$O(N+M)$ (遍历完两个链表)
*   **空间**：$O(1)$ (只改指针，没建新节点)

````python
def mergeTwoLists(list1, list2):
    dummy = ListNode(0) # 裁判起点（虚拟头）
    tail = dummy        # 裁判当前站的位置
    
    while list1 and list2: # 只要两队都还有人
        if list1.val < list2.val:
            # 1. 裁判连上 list1 (注意：直接连节点，不是连值)
            tail.next = list1 
            # 2. 裁判往前走 (走到刚才连上的 list1 节点)
            tail = tail.next
            # 3. list1 队员往前走 (准备下一个候选人)
            list1 = list1.next 
        else:
            # 同理处理 list2
            tail.next = list2      # 连上 list2 节点
            tail = tail.next       # 裁判移到 list2
            list2 = list2.next     # list2 队员前移
    
    # 循环结束，肯定有一队空了，把剩下那队直接连上
    # 这里也不需要 .val，直接把剩下的整条链子挂上去
    if list1:
        tail.next = list1
    else:
        tail.next = list2
        
    return dummy.nextS
````
