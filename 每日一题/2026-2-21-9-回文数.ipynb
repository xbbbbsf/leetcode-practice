字符串方法：

```python
class Solution:
    def isPalindrome(self, x: int) -> bool:
        #  转字符串
        s = str(x)
        
        #  双指针初始化
        left = 0
        right = len(s) - 1  # ⚠️ 注意减1！
        
        #  双指针比对
        while left < right:  # ⚠️ 小于号！
            if s[left] != s[right]:  # ⚠️ 用 == 比较
                return False  # 发现不等，立即返回
            left += 1   # 左指针向右
            right -= 1  # ⚠️ 右指针向左！
        
        #  全部比对完成，是回文
        return True
```

数字方法

```python
class Solution:
    def isPalindrome(self, x: int) -> bool:
        # 1. 负数不是回文
        if x < 0:
            return False
        
        # 2. 保存原数
        original = x
        
        # 3. 反转整数
        reversed_num = 0
        while x != 0:
            pop = x % 10
            x //= 10
            reversed_num = reversed_num * 10 + pop
        
        # 4. 比较原数和反转后的数
        return original == reversed_num
```
